### Запуск из Python

```python
from hydra import compose, initialize
from omegaconf import DictConfig

from fmlib.training import TrainingState, instantiate_state, training_loop
```

```python
with initialize(config_path="configs/train"):
    cfg: DictConfig = compose(config_name="feature_transformer")
```

```python
state: TrainingState = instantiate_state(cfg)
```

```python
results = training_loop(state)
```
