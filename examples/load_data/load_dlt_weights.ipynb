### Десериализация весов

```python
path = "/home/datalab/nfs/camomile-artifacts/DLT/ds_weights/weights.bin"
```

```python
weights = torch.load(path)
```

### Разделение весов по ключам

```python
embedding_model_weights = OrderedDict([(k, v) for k, v in weights.items() if ("base_model" in k)])
```

```python
output_model_weights = OrderedDict([(k, v) for k, v in weights.items() if ("output_layers" in k)])
```

```python
len(embedding_model_weights) + len(output_model_weights), len(weights)
```

### Сохранение всех весов

```python
stt.save_file(weights, "/home/datalab/nfs/amazmefmlib/examples/configs/inference/dlt/weights.safetensors")
```

### Сохранение модели тела (эмбеддингов)

```python
stt.save_file(
    embedding_model_weights, "/home/datalab/nfs/amazmefmlib/examples/configs/inference/dlt_embedding/weights.safetensors"
)
```
